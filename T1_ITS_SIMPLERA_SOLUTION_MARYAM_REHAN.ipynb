```python
from datetime import datetime

VALID_CATEGORIES = {
    "Food",
    "Transport",
    "Utilities",
    "Entertainment",
    "Other"
}


def validate_amount(value):
    """Check whether amount is positive."""
    try:
        amount = float(value)
        return amount > 0
    except (ValueError, TypeError):
        return False


def validate_date(date_str):
    """Check date format YYYY-MM-DD."""
    try:
        datetime.strptime(date_str, "%Y-%m-%d")
        return True
    except (ValueError, TypeError):
        return False


def validate_category(category):
    """Check whether category is valid."""
    return category.strip().title() in VALID_CATEGORIES


print("All validation functions loaded successfully!")
```

```python
# IT Simplera - Expense Analytics Engine

from statistics import mean


def add_expense(expenses, expense_id, date, category, amount, description):
    """Add a new expense record."""

    expense = {
        "id": expense_id,
        "date": date,
        "category": category,
        "amount": float(amount),
        "description": description
    }

    expenses.append(expense)

    return expense


def filter_by_category(expenses, category):
    """Filter expenses by category."""

    return [
        expense for expense in expenses
        if expense["category"].lower() == category.lower()
    ]


def filter_by_date_range(expenses, start_date, end_date):
    """Filter expenses within a date range."""

    return [
        expense for expense in expenses
        if start_date <= expense["date"] <= end_date
    ]


def get_category_totals(expenses):
    """Calculate total spending for each category."""

    category_totals = {}

    for expense in expenses:
        category = expense["category"]

        category_totals[category] = (
            category_totals.get(category, 0)
            + expense["amount"]
        )

    return category_totals


def check_budget_status(spent, budget):
    """Check category budget usage and return status."""

    if budget <= 0:
        return "Invalid Budget"

    percentage = (spent / budget) * 100

    if percentage < 80:
        return "Safe"
    elif percentage <= 100:
        return "Warning"
    else:
        return "Exceeded"


def compute_summary_tuple(expenses):
    """Calculate overall expense summary."""

    if not expenses:
        return (0, 0, 0, 0)

    amounts = [expense["amount"] for expense in expenses]

    return (
        sum(amounts),
        mean(amounts),
        max(amounts),
        min(amounts)
    )


print("Analytics engine created successfully!")
```

```python
# IT Simplera - Main CLI Application

from datetime import date

def run_cli():
    """Run the Personal Expense and Budget CLI."""

    expenses = []
    budgets = {}

    while True:
        print("\n" + "=" * 45)
        print(" PERSONAL EXPENSE & BUDGET ANALYZER")
        print("=" * 45)

        print("1. Add Expense")
        print("2. View All Expenses")
        print("3. Filter by Category")
        print("4. Set Budget")
        print("5. Analytics Report")
        print("6. Exit")

        choice = input("\nEnter your choice: ").strip()

        if choice == "1":
            amount = input("Enter amount: ")
            expense_date = input("Enter date (YYYY-MM-DD): ")
            category = input("Enter category: ").strip().title()
            description = input("Enter description: ")

            if not validate_amount(amount):
                print("Invalid amount!")
                continue

            if not validate_date(expense_date):
                print("Invalid date!")
                continue

            if not validate_category(category):
                print("Invalid category!")
                continue

            add_expense(
                expenses,
                len(expenses) + 1,
                expense_date,
                category,
                float(amount),
                description
            )

            print("Expense added successfully!")

        elif choice == "2":
            if not expenses:
                print("No expenses recorded.")
            else:
                print("\nID | DATE | CATEGORY | AMOUNT | DESCRIPTION")
                print("-" * 65)

                for item in expenses:
                    print(
                        item["id"], "|",
                        item["date"], "|",
                        item["category"], "|",
                        item["amount"], "|",
                        item["description"]
                    )

        elif choice == "3":
            category = input("Enter category: ").strip()

            results = filter_by_category(expenses, category)

            if results:
                for item in results:
                    print(item)
            else:
                print("No matching expenses found.")

        elif choice == "4":
            category = input("Enter category: ").strip().title()
            budget_input = input("Enter monthly budget: ")

            if not validate_category(category):
                print("Invalid category!")
                continue

            if not validate_amount(budget_input):
                print("Invalid budget amount!")
                continue

            budgets[category] = float(budget_input)

            spent = get_category_totals(expenses).get(category, 0)

            status = check_budget_status(
                spent, budgets[category]
            )

            print("Budget saved successfully!")
            print("Current Status:", status)

        elif choice == "5":
            summary = compute_summary_tuple(expenses)

            print("\n========== ANALYTICS REPORT ==========")
            print("Total Spending:", summary[0])
            print("Average Expense:", summary[1])
            print("Highest Expense:", summary[2])
            print("Lowest Expense:", summary[3])

            print("\nCategory Totals:")

            for category, total in get_category_totals(expenses).items():
                print(category, ":", total)

            print("\nBudget Status:")

            totals = get_category_totals(expenses)

            for category, budget in budgets.items():
                spent = totals.get(category, 0)
                print(
                    category,
                    ":",
                    check_budget_status(spent, budget)
                )

        elif choice == "6":
            print("Thank you for using Expense Analyzer!")
            break

        else:
            print("Invalid choice! Please select 1-6.")
```

```python

import unittest

class TestExpenseAnalyzer(unittest.TestCase):

    def test_total_expense(self):
        expenses = [
            {"amount": 1000},
            {"amount": 500}
        ]
        result = compute_summary_tuple(expenses)
        self.assertEqual(result[0], 1500)

    def test_average_expense(self):
        expenses = [
            {"amount": 1000},
            {"amount": 500}
        ]
        result = compute_summary_tuple(expenses)
        self.assertEqual(result[1], 750)

    def test_empty_expense_list(self):
        result = compute_summary_tuple([])
        self.assertEqual(result[0], 0)
        self.assertEqual(result[1], 0)

    def test_highest_expense(self):
        expenses = [
            {"amount": 1000},
            {"amount": 2500},
            {"amount": 500}
        ]
        result = compute_summary_tuple(expenses)
        self.assertEqual(result[2], 2500)

    def test_category_filter(self):
        expenses = [
            {"amount": 1000, "category": "Food"},
            {"amount": 500, "category": "Transport"}
        ]
        result = filter_by_category(expenses, "Food")
        self.assertEqual(len(result), 1)
        self.assertEqual(result[0]["category"], "Food")

    def test_invalid_amount(self):
        self.assertFalse(validate_amount("-100"))
        self.assertFalse(validate_amount("abc"))


unittest.main(argv=[''], exit=False, verbosity=2)
```
